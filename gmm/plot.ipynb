```python
# preamble
import sys
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np
import time

print('Now: ' + time.strftime('%c'))
print(sys.version)

suffixes = ['0',          '1',       '2',       '3',        '4',        '5',        'cs']
descr =    ['nch = 1.25', 'nch = 2', 'nch = 5', 'nch = 10', 'nch = 20', 'nch = 50', 'common sigma']

nfiles = len(suffixes)
names = ['./logs/log-' + s + '.txt' for s in suffixes]

# default colors from T10-palette
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red', 'tab:purple', 'tab:brown', 'tab:pink', 'tab:gray', 'tab:olive', 'tab:cyan']

# load data
data = [None] * nfiles
for i in range(nfiles):
    data[i] = np.loadtxt(names[i], comments='#', delimiter=' ', ndmin=2, usecols = (1,3, 5,7, 9,11, 13))

# common plot parameters
plt.rcParams.update({'font.size': 16})
# maxx = 25000
maxx = -1 # means no limit
```

```python
# hidden model (z)

plt.rcParams["figure.figsize"] = (30,6)
fig, ax = plt.subplots(1, 2)
fig.suptitle('Classes (c)')

for i in range(nfiles):
    ax[0].plot(data[i][:,1],data[i][:,2], color=colors[i]);
ax[0].set_ylabel('clen');
ax[0].grid(True);
if maxx>0: ax[0].set_xlim((0, maxx));

for i in range(nfiles):
    ax[1].plot(data[i][:,1],data[i][:,3], color=colors[i]);
ax[1].set_ylabel('closs');
ax[1].grid(True);
if maxx>0: ax[1].set_xlim((0, maxx));
ax[1].set_ylim((0, 0.8));

ax[1].legend(descr);
```

```python
# Image model

plt.rcParams["figure.figsize"] = (30,6)
fig, ax = plt.subplots(1, 2)
fig.suptitle('Image model (x)')

for i in range(nfiles):
    ax[0].plot(data[i][:,1],data[i][:,4], color=colors[i]);
ax[0].set_ylabel('xlen');
ax[0].grid(True);
if maxx>0: ax[0].set_xlim((0, maxx));

for i in range(nfiles):
    ax[1].plot(data[i][:,1],data[i][:,5], color=colors[i]);
ax[1].set_ylabel('xloss');
ax[1].grid(True);
if maxx>0: ax[1].set_xlim((0, maxx));
ax[1].set_ylim((0.2, 1));

ax[1].legend(descr);
```

```python
plt.rcParams["figure.figsize"] = (20,6)
fig, ax = plt.subplots(1, 1)
fig.suptitle('Inconsistency')

for i in range(nfiles):
    ax.plot(data[i][:,1],data[i][:,6], color=colors[i]);
ax.grid(True);
if maxx>0: ax.set_xlim((0, maxx));
ax.set_ylim((-0.001, 0.02));

ax.legend(descr);
```

```python
# times

for i in range(nfiles):
    num = data[i].size//7
    last_time = data[i][num-1,0]
    last_it = data[i][num-1,1]
    speed = last_it/last_time
    print(descr[i], '(it/sec):' , speed, '\t', last_it, 'iterations', last_time/3600, 'h')
```
